# 01. Embeddings & Similarity Exploration
Test embeddings generation using GoogleGenerativeAIEmbeddings and calculate cosine similarity between candidate resume profile and job postings.

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import GoogleGenerativeAIEmbeddings

load_dotenv()
embeddings = GoogleGenerativeAIEmbeddings(model='gemini-embedding-001')
vec1 = embeddings.embed_query('Machine Learning Engineer with Python skills')
vec2 = embeddings.embed_query('Data Science Intern with Python and SQL')
print('Embedding dimension:', len(vec1))
```
